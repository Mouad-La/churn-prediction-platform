```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine('postgresql://postgres:138322527@localhost:5432/churn_db')
```

```python
df = pd.read_csv("../data/processed/telco_churn_cleaned.csv")
df.head()
```

```python
customers_df = df[['customerID', 'gender', 'SeniorCitizen', 'Partner', 'Dependents', 'tenure']].copy()
customers_df.columns = ['customer_id', 'gender', 'senior_citizen', 'partner', 'dependents', 'tenure']
```

```python
subscriptions_df = df[['customerID', 'Contract', 'InternetService', 'MonthlyCharges', 'TotalCharges', 'Churn']].copy()
subscriptions_df.columns = ['customer_id', 'contract', 'internet_service', 'monthly_charges', 'total_charges', 'churn']
```

```python
customers_df.to_sql('customers', engine, if_exists='append', index=False)
subscriptions_df.to_sql('subscriptions', engine, if_exists='append', index=False)
```
